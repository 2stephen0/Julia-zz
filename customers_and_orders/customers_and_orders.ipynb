**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
# evaluate the contents of the file


from os.path import exists
import csv

datafile = "./sample_data/customers.csv"

if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")


states = set() #Contador de estados sin repetirlo

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')

  next(csvreader)

  for row in csvreader:
    state = row[4].strip().lower()  #Se quitan los espacios y se deja todo en minuscula para reducir el error de tomar un mismo estado teniendo una letra en mayuscula
    states.add(state)

print("Numeros de estados diferentes:", len(states))
```

**Question 3** What is the state with most customers?





```python
# open file and init vars
# evaluate the contents of the file

from os.path import exists
import csv

Contar_estado = {}

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')

  next(csvreader)

  for row in csvreader:
    state = row[4].strip().lower()

    if state in Contar_estado:
      Contar_estado[state] += 1
    else:
      Contar_estado[state] = 1

# encuentra el estado con más clientes
max_state = []
max_count = 0

for state, count in Contar_estado.items():
  if count > max_count:
    max_count = count
    max_state = state

print("El estado con mas clientes es:", max_state, "Con", max_count)
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
# evaluate the contents of the file


from os.path import exists
import csv

Contar_estado = {}

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')

  next(csvreader)

  for row in csvreader:
    state = row[4].strip().lower()

    if state in Contar_estado:
      Contar_estado[state] += 1
    else:
      Contar_estado[state] = 1

# encuentra el estado con menos  clientes

min_state = []
min_count = float('inf') #Toma el numero mas alto

for state, count in Contar_estado.items():
  if count < min_count:
    min_count = count
    min_state = [state]
  elif count == min_count:
    min_state.append(state) #Guarda todos los estados que tienen menos clientes

print("El estado con menos clientes es:", min_state, "Con", min_count)
```

**Question #1**: How many unique orders are in the orders.csv file?

```python
from os.path import exists
import csv

ordenes = set() #Buscar ordenes sin repetir

with open(datafile2, 'r') as fl:
  csvreader2 = csv.reader(fl, delimiter=',')

  next(csvreader2)

  for row in csvreader2:
    orden_id = row[1].strip()
    ordenes.add(orden_id)

print("El numero de ordenes unicas es:", len(ordenes))
```

**Question #2**: What is the average number of items per order (rounded to two decimal places)?

```python
from os.path import exists
import csv

ordenes = {}

with open(datafile2, 'r') as fl:
  csvreader2 = csv.reader(fl, delimiter=',')

  next(csvreader2)

  for row in csvreader2:
    try:
      orden_id = int(float(row[1]))  # limpiar datos

      if orden_id in ordenes:
       ordenes[orden_id] += 1
      else:
        ordenes[orden_id] = 1

    except:
      continue

# calcular promedio
total_items = sum(ordenes.values())
total_ordenes = len(ordenes)

promedio = total_items / total_ordenes

print("El promedio de items por pedido es",round(promedio, 2))
```

**Question #3**: What is the highest number of items per order?

```python
from os.path import exists
import csv

ordenes = {}

with open(datafile2, 'r') as fl:
  csvreader2 = csv.reader(fl, delimiter=',')

  next(csvreader2)

  for row in csvreader2:
    try:
      orden_id = int(float(row[1]))  # limpiar datos

      if orden_id in ordenes:
       ordenes[orden_id] += 1
      else:
        ordenes[orden_id] = 1

    except:
      continue

  max_items = max(ordenes.values())
print("El numero maximo de items por pedido es:", max_items)
```

**Question #4**: What is the number of orders placed in October 2021?

```python
from os.path import exists
import csv

ordenes_octubre = set()

with open(datafile2, 'r') as fl:
  csvreader2 = csv.reader(fl, delimiter=',')

  next(csvreader2)

  for row in csvreader2:
      try:
        orden_id = int(float(row[1]))
        date = row[2].strip()

        if date.startswith("2021-10"):
          ordenes_octubre.add(orden_id)

      except:
        continue

print("El numero de ordenes en octubre del 2021 es: ",len(ordenes_octubre))
```

**Question #5**: Which customer spent the most amount of money in 2021?

```python
from os.path import exists
import csv

cliente = {}

#nombre del cliente
with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl)
  next(csvreader)

  for row in csvreader:
    cliente_id = int(row[0])
    name = row[1] + " " + row[2]
    cliente[cliente_id] = name

#Gasto por cliente en 2021
gasto_cliente = {}
procesar_orden = set()

with open(datafile2, 'r') as fl:
  csvreader2= csv.reader(fl)
  next(csvreader2)

  for row in csvreader2:
    try:
      cliente_id = int(row[0])
      orden_id = int(float(row[1]))
      date = row[2]
      orden_total = float(row[3])

      if not date.startswith("2021"):
        continue

      if orden_id in procesar_orden:
        continue

      procesar_orden.add(orden_id)

      if cliente_id in gasto_cliente:
        gasto_cliente[cliente_id] += orden_total
      else:
        gasto_cliente[cliente_id] = orden_total

    except:
      continue

#  3. Encontrar el que más gastó
max_cliente = []
max_gasto = 0

for clientes, total in gasto_cliente.items():
  if total > max_gasto:
    max_gasto = total
    max_cliente = clientes


print("Cliente ID:", max_cliente,
      "\nNombre:", cliente.get(max_cliente),
      "\nTotal Gastado:", round(max_gasto, 2))
```

**Question #6**: Historically, what is the best month for sales?

```python
from os.path import exists
import csv

ventas_mes = {}
procesar_orden = set()

with open(datafile2, 'r') as fl:
  csvreader = csv.reader(fl)
  next(csvreader)

  for row in csvreader:
    try:
      orden_id = int(float(row[1]))
      date = row[2].strip()
      total = float(row[3])

      if date == "": #Salta los espacios vacios en las fechas
        continue

      if orden_id in procesar_orden:
        continue

      procesar_orden.add(orden_id)

      mes = date[:7]

      if mes in ventas_mes:
        ventas_mes[mes] += total
      else:
        ventas_mes[mes] = total

    except:
      continue

mejor_mes = None
max_ventas = 0

for mes, total in ventas_mes.items():
  if total > max_ventas:
    max_ventas = total
    mejor_mes = mes

print("Mes:", mejor_mes,
      "\nVentas:", round(max_ventas, 2))
```
